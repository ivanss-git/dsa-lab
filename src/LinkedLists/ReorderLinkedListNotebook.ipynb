# **Reorder Linked List**

### How does this utilize the fast and slow pointer (Hare and Tortoise) pattern in its purest sense?
- This utilizes the fast and slow pointer pattern by finding the middle of the list using <FAST> and <SLOW> pointers, spliting the list into two halves. We then reverse the second half of the list so it is easy to merge nodes from the front and back alternately. Lastly, we merge the two halves together one-by-one, beginning with one node from the first half then one from the traversed second half and repeat. 

### What are the edge cases with even versus odd length lists?
- The length of the list changes how you identify the split point and how you determine when to merge. For odd length lists, the fast.next pointer stops when fast.next is null and the slow pointer lands on the middle node. The first half gets an odd number of nodes and the second half gets an even number of nodes. For lists of even lengths, the fast pointer stops when fast itself is null and the slow pointer lands on the second middle node. Both halves get an even number of nodes. If you do not explicitly break the link between both halves, you will create a ciurcular loop. To avoid this, you set slow.next = null right after finding the middle to ensure that the first half has a clean tail.

### What is the mathematical proff that the slow pointer will reach exactly the midpoint when the fast pointer reaches the end?
- Instead of of looking at the positions as moving forward, think of the pointers as shrinking the remaining lists from both ends. Everytime you advance a pointer, you remove 1/2 nodes from the left/right side of the unvisited list. Because the fast pointer visits nodes twice as fast, the total unvisited distance between them is 1 node per step. Ex: slow is at index k, and fast is at index 2k. Slow moves to k+1 and fast moves to 2(k+1), which is equal to 2k + 2. Because the multiplier is preserved at each step, the final stopping point of slow is guaranteed to be exactly half of the final boundary that fast reaches. 

### Why can't you just use an array instead?
- You CAN use an array, but it changes the algorithmic efficiency and presents a tradeoff between time complexity (1 pass vs 3 passes), space complexity (creating an array requires memory), and code simplicity (array provides simpler indexing). The pointer approach is preferred because it avoids having to copy all node references into an array, using zero extra memory.

## Pattern Questions

### What problem is this pattern designed to solve? 
- This pattern is designed to solve structural reodering where nodes from the front and back of a list must be interleaved. It uses a combination of pointers and list manipulation patterns to split the linked list into two distinct list, reverse the second half of the list to enable backward traversal, and merge the first half with the reversed second half by weaving their nodes together.

### Why is it better than brute force?
- This approach is better than the brute force approach because it avoids having to store all nodes in an array and alternating between nodes to reshape a list into the required order.

### What data structures are commonly used with it?
- The data structures commonly associated with this question involve a stack and recursion, with the most optimal implementation involving two pointers and a linked list. 

## Key Takeaways
- The 3 step blueprint: The most optimal apprach relies on a linear, 3-step algorithm
    * Find the middle using slow and fast pointers to split the list into two halves.
    * Reverse the second half in place
    * Merge the two halves by interleaving nodes alternately.

- Complexity: This results in an optimal O(n) time complexity and O(1) space complexity.

- Structural Isomorphis: The n-1 logic of recursion maps perfectly to linked lists because both structures share a linear, one-directional topology. 
    * In recursion, you process a sequence by stripping away one element and passing the remaining chain down
    * In linked lists, the process involves reading the current node and passing the remaining sublist down.
    * Steps: Base case, recursive step, multiplier.
