# Linked List Implementations Discussion

**Linked List Cycle Detection**

Why are you guaranteed to catch the cycle? How do you prove the fast pointer won't just skip over the slow pointer?

- The relative speed between the two pointers is exactly **1 node per iteration**. Once both pointers are inside the cycle, the distance between them decreases by 1 each iteration. Eventually, that distance becomes 0, meaning the fast pointer must meet the slow pointer—it cannot skip over it.

What happens if the fast pointer moved 3 steps at a time instead of 2? Does it still work? Is it better?

- It can still detect a cycle, but it is not necessarily better. Moving 3 steps changes the relative speed to 2 nodes per iteration. While a meeting is still possible, it depends on the cycle length, and the behavior is less straightforward than Floyd's standard 2-step approach. Using 2 steps is simpler, well-proven, and guarantees detection for any cycle length.

If the **O(1)** memory constraint didn't matter, how could you solve this by modifying the actual list node object as you traverse it?

- You could add a `boolean visited` field to the `ListNode` class and mark each node as you visit it. If you reach a node that has already been marked as visited, a cycle exists. Alternatively, you could temporarily change a node's value to a unique sentinel value (e.g., `Integer.MAX_VALUE`) and check whether you encounter that value again.

What problem is this pattern designed to solve? 

- This pattern is designed to determine whether a data structure contains an infinite loop, meaning a node's pointer references a previously visited node instesd of ending at `null`. Without this pattern, a standard traversal algorithm would get trapped in the loop and run forever, causing a crash or hang.

Why is it better than brute force?

- This solution is better than brute force because it ensures that we only visit each node once while also making sure that memory usage does not increase as the sise up the input increases (`O(n)` time comp `O(1)` space comp)

What data structures are commonly used with it?

- Data strucutures commonly associated with this problem are HashSets, LinkedLists, and Two pointers. The naive approach would be to use a hashSet, which takes `O(1)` time to detect duplicates, however it requires `O(n)` extra space. The optimal solution involves moving two pointers at different speeds through a continuous loops, reducing the gap between pointers with each step. This continues until the fast pointer catches up to the slow pointer, confirming the cycle.
